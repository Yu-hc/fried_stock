```python
import tejapi
import os
api_key = os.getenv("TEJ_API_KEY")
tejapi.ApiConfig.api_key = api_key
tejapi.ApiConfig.ignoretz = True
```

```python
# Search for tables containing "Concentration" or "Chip" keywords
search_results = tejapi.search_table("籌碼")

# Print the table codes and descriptions
for table in search_results:
       print(f"{table['tableCode']} - {table['name']}")
```
