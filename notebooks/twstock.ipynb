```python
import twstock

# 取得 2330 台積電的即時報價
realtime_data = twstock.realtime.get('2330')
if realtime_data['success']:
       print("目前價格:", realtime_data['realtime']['accumulate_trade_volume'])
       print(realtime_data['realtime'])
```

```python


def get_current_price(stock_data):
       # 1. 優先使用最新成交價
       price = stock_data.get('latest_trade_price')
       if price and price != '-':
              return float(price)
       
       # 2. 如果沒有成交價，嘗試使用最佳委賣價的第一檔 (排除 0.0000 市價單)
       ask_prices = [p for p in stock_data.get('best_ask_price', []) if p != '-' and float(p) > 0]
       if ask_prices:
              return float(ask_prices[0]) # 拿最便宜想賣的價格 (本例為 24.3)

       # 3. 如果連委賣價都沒有，嘗試使用最佳委買價的第一檔
       bid_prices = [p for p in stock_data.get('best_bid_price', []) if p != '-' and float(p) > 0]
       if bid_prices:
              return float(bid_prices[0])

       # 4. 真的都沒資料，回傳開盤價當作參考
       return float(stock_data.get('open', 0))

current_price = get_current_price(realtime_data['realtime'])
print(f"估算當前股價為: {current_price}") 
# 輸出: 估算當前股價為: 24.3
```
